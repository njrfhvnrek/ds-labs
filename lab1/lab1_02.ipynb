```python
from math import isqrt


def is_prime(n: int) -> bool:
    """Проверяет, является ли число n простым.

    Используется наивная проверка делителями от 2 до sqrt(n).
    Для n <= 1 возвращает False (по определению простого числа).

    Аргументы:
        n (int): проверяемое целое число.

    Возвращает:
        bool: True, если n — простое; False в противном случае.
    """
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    # Достаточно проверить нечётные делители до sqrt(n) включительно.
    # Если у n есть делитель d > sqrt(n), то парный ему делитель n/d < sqrt(n)
    # уже был бы найден на предыдущих итерациях.
    limit = isqrt(n)
    for d in range(3, limit + 1, 2):
        if n % d == 0:
            return False
    return True
```

Демонстрация работы


```python
# Демонстрация работы функции на примерах из задания
print("Проверка простоты чисел")
print("-" * 30)

# Пример 1: n = 29 (простое)
print(f"is_prime(29) -> {is_prime(29)}")

# Пример 2: n = 25 (составное, 25 = 5 * 5)
print(f"is_prime(25) -> {is_prime(25)}")

# Дополнительные примеры для наглядности
print(f"is_prime(2)  -> {is_prime(2)}   # наименьшее простое")
print(f"is_prime(1)  -> {is_prime(1)}  # 1 не является простым")
print(f"is_prime(97) -> {is_prime(97)}  # простое")
print(f"is_prime(100)-> {is_prime(100)} # составное")
```
