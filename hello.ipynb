```python
def greet(name: str) -> str:
    """Возвращает приветствие для указанного имени."""
    return f"Привет, {name}!"

def compute_statistics(numbers):
    """Считает базовые статистики по списку чисел."""
    if not numbers:
      return None
    return {
      "count": len(numbers),
      "sum": sum(numbers),
      "min": min(numbers),
      "max": max(numbers),
      "avg": sum(numbers) / len(numbers),
    }
```

```python
# 1. Приветствие
print(greet("Студент"))
# 2. Работа со списком
sample = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
stats = compute_statistics(sample)
print(f"Список: {sample}")
for key, value in stats.items():
    print(f" {key:>5}: {value}")
```
